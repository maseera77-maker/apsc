## Example 10 — Password Strength

| Stream | Task |
|---|---|
| 🖥️ **BCA** | Check: 'hi', 'hello123', 'Hello@123', 'a1'. Which is strong? |
| 📊 **BBA** | Policy compliance checker. What rules does your organisation use? |

> 🔍 *Common thinking gap:* Cannot combine AND; checks conditions separately.

```python
# Example 10 — Password Strength
password = "Hello@123"   # Try: "hi", "hello123", "Hello@123", "a1"

# MY PREDICTION: _____________

long_enough  = len(password) >= 8
has_digit    = any(c.isdigit() for c in password)
has_upper    = any(c.isupper() for c in password)

if long_enough and has_digit and has_upper:
    print("✅ Strong password")
elif long_enough and has_digit:
    print("⚠️ Medium — add an uppercase letter")
else:
    print("❌ Weak password — too short or missing digits")

print(f"  Length OK: {long_enough} | Has digit: {has_digit} | Has uppercase: {has_upper}")

# ── YOUR TASK ──────────────────────────────────────────
# BCA: Add a 4th rule: has_special = any(c in "!@#$%^&*" for c in password)
#      Update the if-condition to include it.
# BBA: What are 3 password rules your college/bank enforces? Write them as comments.

# ✏️ Reflection 10: What surprised you? What was the bug?
reflection_10 = "Type your one-sentence reflection here"
print(reflection_10)
```
